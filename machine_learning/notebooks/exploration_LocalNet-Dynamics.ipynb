```python
import pandas as pd
import numpy as np

df = pd.read_csv("../../data/localnet_dynamics.csv")

print("Dimensions :", df.shape)

print("\nColonnes :")
print(df.columns.tolist())

print("\nPremières lignes :")
display(df.head())

print("\nTypes des colonnes :")
print(df.dtypes)
```

```python
# 1. Valeurs manquantes
print("===== VALEURS MANQUANTES =====")

missing = pd.DataFrame({
    "Valeurs manquantes": df.isnull().sum(),
    "Pourcentage (%)": (df.isnull().mean() * 100).round(2)
})

display(missing[missing["Valeurs manquantes"] > 0]
        .sort_values("Pourcentage (%)", ascending=False))


# 2. Doublons
print("\n===== DOUBLONS =====")

print("Nombre de lignes dupliquées :", df.duplicated().sum())


# 3. Sessions de collecte
print("\n===== SESSIONS =====")

print("Nombre de sessions :", df["run_id"].nunique())

print("\nNombre d'observations par session :")
display(df["run_id"].value_counts())


# 4. Équipements et destinations
print("\n===== ÉQUIPEMENTS =====")

print(df["host_label"].value_counts())

print("\n===== DESTINATIONS =====")

print(df["target"].value_counts())


# 5. Période couverte
print("\n===== PÉRIODE =====")

dates = pd.to_datetime(df["timestamp_utc"], utc=True)

print("Première mesure :", dates.min())
print("Dernière mesure :", dates.max())
print("Durée :", dates.max() - dates.min())
```

```python
# Conversion du timestamp
df["timestamp_utc"] = pd.to_datetime(
    df["timestamp_utc"], utc=True
)

# Tri chronologique
df = df.sort_values("timestamp_utc").reset_index(drop=True)

# Calcul des intervalles réels
df["intervalle_reel"] = df["timestamp_utc"].diff().dt.total_seconds()

print("===== DISTRIBUTION DES INTERVALLES =====")

print(df["intervalle_reel"].value_counts().head(15))

print("\n===== STATISTIQUES DES INTERVALLES =====")

display(df["intervalle_reel"].describe())

# Intervalles différents de 30 secondes
print("\n===== INTERVALLES DIFFERENTS DE 30 SECONDES =====")

print(
    "Nombre :", 
    (df["intervalle_reel"] != 30).sum() - 1
)

# Intervalles supérieurs à 30 secondes
gaps = df[df["intervalle_reel"] > 30]

print("\nNombre d'intervalles supérieurs à 30 secondes :", len(gaps))

print("\nExemples :")
display(
    gaps[["timestamp_utc", "intervalle_reel", "gap_before_s"]].head(15)
)

# Comparaison avec la colonne interval_s
print("\n===== INTERVAL_S =====")
print(df["interval_s"].value_counts(dropna=False))
```

```python
# 1. Répartition des intervalles par catégorie
conditions = [
    df["intervalle_reel"].between(25, 35),
    df["intervalle_reel"].between(36, 45),
    df["intervalle_reel"].between(46, 75),
    df["intervalle_reel"] > 75
]

categories = [
    "25 à 35 secondes",
    "36 à 45 secondes",
    "46 à 75 secondes",
    "Plus de 75 secondes"
]

df["categorie_intervalle"] = np.select(
    conditions,
    categories,
    default="Autre"
)

print("===== CATEGORISATION DES INTERVALLES =====")
print(df["categorie_intervalle"].value_counts())


# 2. Vérifier la cohérence de gap_before_s
print("\n===== COHERENCE GAP_BEFORE_S =====")

comparaison = (
    df["intervalle_reel"] - df["gap_before_s"]
).abs()

print("Écarts entre les deux colonnes :", (comparaison > 0).sum())


# 3. Vérifier la continuité de sample_index
print("\n===== CONTINUITE SAMPLE_INDEX =====")

diff_index = df["sample_index"].diff()

print("Nombre de sauts dans sample_index :",
      (diff_index > 1).sum())

print("Nombre de retours en arrière :",
      (diff_index < 1).sum())


# 4. Examiner les interruptions les plus longues
print("\n===== PLUS GRANDES INTERRUPTIONS =====")

display(
    df.nlargest(15, "intervalle_reel")[
        ["sample_index", "timestamp_utc",
         "intervalle_reel", "gap_before_s"]
    ]
)
```

```python
# Variables réseau principales
variables = [
    "rtt_avg_ms",
    "rtt_min_ms",
    "rtt_max_ms",
    "rtt_median_ms",
    "rtt_stddev_ms",
    "jitter_ms",
    "packet_loss_pct",
    "gateway_rtt_avg_ms",
    "gateway_jitter_ms"
]

print("===== STATISTIQUES DES VARIABLES RESEAU =====")
display(df[variables].describe().T)

# Vérification des valeurs négatives
print("\n===== VALEURS NEGATIVES =====")

for col in variables:
    print(col, ":", (df[col] < 0).sum())

# Distribution des labels
labels = [
    "label_outage",
    "label_packet_loss",
    "label_high_latency",
    "label_high_jitter",
    "label_gateway_degraded",
    "label_anomaly"
]

print("\n===== DISTRIBUTION DES LABELS =====")

for col in labels:
    print(f"\n{col}")
    print(df[col].value_counts(dropna=False))

# Répartition des anomalies
print("\n===== POURCENTAGE DES ANOMALIES =====")

for col in labels:
    print(
        col,
        ":",
        round(df[col].mean() * 100, 2),
        "%"
    )
```

```python
# 1. Examiner les 18 observations sans mesures RTT
print("===== OBSERVATIONS SANS RTT =====")

display(
    df[df["rtt_avg_ms"].isna()][
        ["sample_index", "timestamp_utc",
         "packet_loss_pct", "label_outage",
         "label_anomaly"]
    ]
)

# 2. Comparer les mesures selon les labels
print("\n===== MESURES SELON LES LABELS =====")

for label, variable in [
    ("label_high_latency", "rtt_avg_ms"),
    ("label_high_jitter", "jitter_ms"),
    ("label_packet_loss", "packet_loss_pct"),
    ("label_gateway_degraded", "gateway_rtt_avg_ms"),
    ("label_anomaly", "rtt_avg_ms")
]:
    print(f"\n{label} / {variable}")

    display(
        df.groupby(label)[variable]
        .agg(["count", "min", "median", "mean", "max"])
    )

# 3. Examiner les valeurs de perte de paquets
print("\n===== DISTRIBUTION DE LA PERTE DE PAQUETS =====")

print(df["packet_loss_pct"].value_counts().head(15))

# 4. Vérifier les relations entre labels
print("\n===== CORRELATION ENTRE LABELS =====")

display(df[labels].corr().round(2))
```

```python
# 1. Nombre de paquets envoyés et reçus
print("===== PAQUETS ENVOYÉS / REÇUS =====")

display(
    df[["packets_sent", "packets_received"]]
    .describe()
)

print("\nValeurs les plus fréquentes :")
print(df["packets_sent"].value_counts().head(10))
print(df["packets_received"].value_counts().head(10))


# 2. Vérifier la cohérence du calcul de perte
print("\n===== VÉRIFICATION DU TAUX DE PERTE =====")

df_verification = df[
    (df["packets_sent"] > 0) &
    df["packets_received"].notna()
].copy()

df_verification["perte_calculee"] = (
    (df_verification["packets_sent"] -
     df_verification["packets_received"])
    / df_verification["packets_sent"] * 100
)

print(
    "Nombre de différences :",
    (
        (df_verification["perte_calculee"] -
         df_verification["packet_loss_pct"]).abs() > 0.01
    ).sum()
)


# 3. Examiner la distinction entre perte et panne
print("\n===== PERTE DE PAQUETS ET PANNE =====")

display(
    pd.crosstab(
        df["label_outage"],
        df["label_packet_loss"],
        margins=True
    )
)

print("\nObservations avec 100 % de perte :")

display(
    df[df["packet_loss_pct"] == 100][
        [
            "timestamp_utc",
            "packets_sent",
            "packets_received",
            "packet_loss_pct",
            "label_outage",
            "label_packet_loss"
        ]
    ]
)
```

```python
# Extraire les observations de panne
outages = df.loc[
    df["label_outage"] == 1,
    ["sample_index", "timestamp_utc", "packet_loss_pct",
     "packets_received"]
].copy()

# Identifier les épisodes consécutifs
outages["nouvel_episode"] = (
    outages["sample_index"].diff().ne(1)
)

outages["episode"] = outages["nouvel_episode"].cumsum()

# Résumer chaque épisode
episodes = outages.groupby("episode").agg(
    debut=("timestamp_utc", "min"),
    fin=("timestamp_utc", "max"),
    nombre_mesures=("sample_index", "count"),
    paquets_recus_total=("packets_received", "sum")
)

print("===== ÉPISODES DE PANNE =====")
display(episodes)

print("Nombre total d'épisodes :", len(episodes))
print("Nombre total de mesures en panne :", len(outages))
```
